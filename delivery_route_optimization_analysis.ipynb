```python
# ============================================================================
# COMPREHENSIVE DATA PROFILING & QUALITY ASSESSMENT
# ============================================================================

print("🔍 ADVANCED DATA SCIENCE PIPELINE")
print("=" * 60)

def comprehensive_data_profiling(delivery_points, vehicles):
    """
    Perform comprehensive data profiling and quality assessment.
    """
    print("📊 1. DATA PROFILING & QUALITY ASSESSMENT")
    print("-" * 50)
    
    # Data shape and types
    print(f"📦 Delivery Points Dataset:")
    print(f"  Shape: {delivery_points.shape}")
    print(f"  Memory usage: {delivery_points.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print(f"  Data types: {delivery_points.dtypes.value_counts().to_dict()}")
    
    print(f"\n🚛 Vehicles Dataset:")
    print(f"  Shape: {vehicles.shape}")
    print(f"  Memory usage: {vehicles.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print(f"  Data types: {vehicles.dtypes.value_counts().to_dict()}")
    
    # Missing values analysis
    delivery_missing = delivery_points.isnull().sum()
    vehicles_missing = vehicles.isnull().sum()
    
    print(f"\n🚨 Missing Values Analysis:")
    print(f"  Delivery Points: {delivery_missing.sum()} total missing")
    print(f"  Vehicles: {vehicles_missing.sum()} total missing")
    
    # Duplicate analysis
    delivery_duplicates = delivery_points.duplicated().sum()
    vehicles_duplicates = vehicles.duplicated().sum()
    
    print(f"\n🔄 Duplicate Analysis:")
    print(f"  Delivery Points duplicates: {delivery_duplicates}")
    print(f"  Vehicles duplicates: {vehicles_duplicates}")
    
    # Statistical summary
    print(f"\n📈 Statistical Summary - Delivery Points:")
    numeric_cols = delivery_points.select_dtypes(include=[np.number]).columns
    stats_summary = delivery_points[numeric_cols].describe()
    display(stats_summary)
    
    return {
        'delivery_missing': delivery_missing,
        'vehicles_missing': vehicles_missing,
        'delivery_duplicates': delivery_duplicates,
        'vehicles_duplicates': vehicles_duplicates,
        'stats_summary': stats_summary
    }

# Outlier detection using IQR method
def detect_outliers_iqr(df, columns):
    """
    Detect outliers using Interquartile Range method.
    """
    outliers_info = {}
    
    for col in columns:
        if df[col].dtype in ['int64', 'float64']:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            
            outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
            outliers_info[col] = {
                'count': len(outliers),
                'percentage': (len(outliers) / len(df)) * 100,
                'lower_bound': lower_bound,
                'upper_bound': upper_bound,
                'outlier_values': outliers[col].tolist()
            }
    
    return outliers_info

# Advanced statistical tests
def perform_advanced_statistical_tests(df):
    """
    Perform advanced statistical tests on the data.
    """
    print("🧪 ADVANCED STATISTICAL TESTING")
    print("-" * 40)
    
    results = {}
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    
    for col in numeric_cols:
        if len(df[col].dropna()) > 3:
            # Normality test
            shapiro_stat, shapiro_p = shapiro(df[col].dropna().sample(min(5000, len(df))))
            
            # Skewness and kurtosis
            skewness = df[col].skew()
            kurtosis = df[col].kurtosis()
            
            results[col] = {
                'shapiro_stat': shapiro_stat,
                'shapiro_p_value': shapiro_p,
                'is_normal': shapiro_p > 0.05,
                'skewness': skewness,
                'kurtosis': kurtosis,
                'distribution_type': 'normal' if abs(skewness) < 0.5 else 'skewed'
            }
            
            print(f"📊 {col}:")
            print(f"  Normality (Shapiro-Wilk): p={shapiro_p:.6f} ({'Normal' if shapiro_p > 0.05 else 'Not Normal'})")
            print(f"  Skewness: {skewness:.3f} ({results[col]['distribution_type']})")
            print(f"  Kurtosis: {kurtosis:.3f}")
            print()
    
    return results

# Perform comprehensive data profiling
profiling_results = comprehensive_data_profiling(delivery_points, vehicles)

# Detect outliers
print("\n🚨 OUTLIER DETECTION")
print("-" * 30)
outlier_cols = ['package_weight', 'package_volume', 'service_time']
outliers_info = detect_outliers_iqr(delivery_points, outlier_cols)

for col, info in outliers_info.items():
    print(f"📦 {col}:")
    print(f"  Outliers: {info['count']} ({info['percentage']:.1f}%)")
    print(f"  Valid range: [{info['lower_bound']:.2f}, {info['upper_bound']:.2f}]")

# Advanced statistical tests
statistical_results = perform_advanced_statistical_tests(delivery_points)

print("✅ Data Profiling Complete!")
```

```python
# ============================================================================
# ADVANCED FEATURE ENGINEERING
# ============================================================================

print("\n⚙️ ADVANCED FEATURE ENGINEERING")
print("=" * 50)

def create_advanced_features(delivery_points, vehicles, distance_matrix):
    """
    Create advanced features for machine learning models.
    """
    df_enhanced = delivery_points.copy()
    
    print("🔧 Creating Geographic Features...")
    
    # 1. GEOGRAPHIC FEATURES
    center_lat, center_lon = -23.5505, -46.6333
    
    # Distance to city center
    df_enhanced['distance_to_center'] = np.sqrt(
        (df_enhanced['latitude'] - center_lat)**2 + 
        (df_enhanced['longitude'] - center_lon)**2
    ) * 111  # Convert to km
    
    # Radial zones (distance bands)
    df_enhanced['zone'] = pd.cut(df_enhanced['distance_to_center'], 
                                bins=[0, 5, 10, 15, np.inf], 
                                labels=['inner', 'middle', 'outer', 'far'])
    
    # Quadrant classification
    df_enhanced['quadrant'] = np.where(df_enhanced['latitude'] > center_lat,
                                      np.where(df_enhanced['longitude'] > center_lon, 'NE', 'NW'),
                                      np.where(df_enhanced['longitude'] > center_lon, 'SE', 'SW'))
    
    print("📦 Creating Package Features...")
    
    # 2. PACKAGE FEATURES
    # Package density (weight/volume ratio)
    df_enhanced['package_density'] = df_enhanced['package_weight'] / df_enhanced['package_volume']
    
    # Package size categories
    df_enhanced['weight_category'] = pd.cut(df_enhanced['package_weight'], 
                                          bins=[0, 5, 15, 25, np.inf],
                                          labels=['light', 'medium', 'heavy', 'very_heavy'])
    
    df_enhanced['volume_category'] = pd.cut(df_enhanced['package_volume'],
                                          bins=[0, 0.5, 1.0, 1.5, np.inf],
                                          labels=['small', 'medium', 'large', 'very_large'])
    
    # Combined size score
    weight_score = df_enhanced['package_weight'] / df_enhanced['package_weight'].max()
    volume_score = df_enhanced['package_volume'] / df_enhanced['package_volume'].max()
    df_enhanced['package_complexity'] = (weight_score + volume_score) / 2
    
    print("⏰ Creating Time Features...")
    
    # 3. TIME FEATURES
    # Time window duration
    df_enhanced['time_window_duration'] = df_enhanced['time_window_end'] - df_enhanced['time_window_start']
    
    # Time window flexibility
    df_enhanced['time_flexibility'] = df_enhanced['time_window_duration'] / 10  # Normalized
    
    # Peak time classification
    def classify_time_period(start_hour):
        if 8 <= start_hour <= 10:
            return 'morning_rush'
        elif 11 <= start_hour <= 14:
            return 'midday'
        elif 15 <= start_hour <= 17:
            return 'afternoon_rush'
        else:
            return 'off_peak'
    
    df_enhanced['time_period'] = df_enhanced['time_window_start'].apply(classify_time_period)
    
    # Service time efficiency
    df_enhanced['service_efficiency'] = df_enhanced['package_weight'] / df_enhanced['service_time']
    
    print("🏢 Creating Business Features...")
    
    # 4. BUSINESS FEATURES
    # Priority encoding
    priority_map = {'low': 1, 'medium': 2, 'high': 3}
    df_enhanced['priority_score'] = df_enhanced['priority'].map(priority_map)
    
    # Delivery difficulty score (combination of factors)
    df_enhanced['delivery_difficulty'] = (
        df_enhanced['priority_score'] * 0.3 +
        df_enhanced['package_complexity'] * 0.4 +
        (1 / df_enhanced['time_flexibility']) * 0.3
    )
    
    print("🌐 Creating Network Features...")
    
    # 5. NETWORK FEATURES
    # Average distance to other points
    avg_distances = []
    for i in range(len(df_enhanced)):
        distances = distance_matrix[i]
        avg_dist = np.mean(distances[distances > 0])  # Exclude self-distance
        avg_distances.append(avg_dist)
    
    df_enhanced['avg_distance_to_others'] = avg_distances
    
    # Isolation score (how far from other deliveries)
    df_enhanced['isolation_score'] = df_enhanced['avg_distance_to_others'] / np.mean(avg_distances)
    
    # Nearest neighbor distance
    nearest_distances = []
    for i in range(len(df_enhanced)):
        distances = distance_matrix[i]
        nearest_dist = np.min(distances[distances > 0])
        nearest_distances.append(nearest_dist)
    
    df_enhanced['nearest_neighbor_distance'] = nearest_distances
    
    print("📊 Creating Statistical Features...")
    
    # 6. STATISTICAL FEATURES
    # Z-scores for key variables
    for col in ['package_weight', 'package_volume', 'service_time']:
        df_enhanced[f'{col}_zscore'] = (df_enhanced[col] - df_enhanced[col].mean()) / df_enhanced[col].std()
    
    # Percentile ranks
    for col in ['package_weight', 'package_volume']:
        df_enhanced[f'{col}_percentile'] = df_enhanced[col].rank(pct=True)
    
    print("🔄 Creating Interaction Features...")
    
    # 7. INTERACTION FEATURES
    # Distance * Priority interaction
    df_enhanced['distance_priority_interaction'] = df_enhanced['distance_to_center'] * df_enhanced['priority_score']
    
    # Weight * Volume interaction
    df_enhanced['weight_volume_interaction'] = df_enhanced['package_weight'] * df_enhanced['package_volume']
    
    # Time * Complexity interaction
    df_enhanced['time_complexity_interaction'] = df_enhanced['service_time'] * df_enhanced['package_complexity']
    
    print("✅ Feature Engineering Complete!")
    print(f"📊 Original features: {len(delivery_points.columns)}")
    print(f"📈 Enhanced features: {len(df_enhanced.columns)}")
    print(f"🆕 New features created: {len(df_enhanced.columns) - len(delivery_points.columns)}")
    
    return df_enhanced

# Apply feature engineering
enhanced_delivery_data = create_advanced_features(delivery_points, vehicles, distance_matrix)

# Display new features summary
print("\n📋 NEW FEATURES SUMMARY:")
new_features = [col for col in enhanced_delivery_data.columns if col not in delivery_points.columns]
for feature in new_features:
    if enhanced_delivery_data[feature].dtype in ['int64', 'float64']:
        print(f"  {feature}: mean={enhanced_delivery_data[feature].mean():.3f}, std={enhanced_delivery_data[feature].std():.3f}")
    else:
        print(f"  {feature}: {enhanced_delivery_data[feature].dtype} with {enhanced_delivery_data[feature].nunique()} unique values")

# Feature correlation analysis
print("\n🔗 FEATURE CORRELATION ANALYSIS:")
numeric_features = enhanced_delivery_data.select_dtypes(include=[np.number]).columns
correlation_matrix = enhanced_delivery_data[numeric_features].corr()

# Find highly correlated features
high_corr_pairs = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_val = correlation_matrix.iloc[i, j]
        if abs(corr_val) > 0.7:  # High correlation threshold
            high_corr_pairs.append((correlation_matrix.columns[i], correlation_matrix.columns[j], corr_val))

print(f"🚨 High correlation pairs (>0.7):")
for feat1, feat2, corr in high_corr_pairs:
    print(f"  {feat1} <-> {feat2}: {corr:.3f}")

print(f"\n✅ Enhanced dataset ready for modeling!")
```

```python
# ============================================================================
# TIME SERIES FORECASTING WITH ADVANCED MODELS
# ============================================================================

print("\n📈 TIME SERIES FORECASTING & DEMAND PREDICTION")
print("=" * 60)

def generate_realistic_demand_series():
    """
    Generate realistic time series data with multiple patterns.
    """
    print("🎲 Generating realistic demand time series...")
    
    # Generate 2 years of daily data
    dates = pd.date_range(start='2022-01-01', end='2023-12-31', freq='D')
    n_days = len(dates)
    
    # Base demand with growth trend
    base_demand = 80 + np.linspace(0, 40, n_days)  # Growing from 80 to 120
    
    # Seasonal components
    yearly_cycle = 25 * np.sin(2 * np.pi * np.arange(n_days) / 365.25)  # Annual
    weekly_cycle = 15 * np.sin(2 * np.pi * np.arange(n_days) / 7)       # Weekly
    monthly_cycle = 10 * np.sin(2 * np.pi * np.arange(n_days) / 30.44)  # Monthly
    
    # Holiday effects (Black Friday, Christmas, etc.)
    holiday_effects = np.zeros(n_days)
    holiday_dates = [60, 120, 180, 240, 300, 330, 425, 485, 545, 605, 665, 695]  # Various holidays
    
    for holiday in holiday_dates:
        if holiday < n_days:
            for i in range(max(0, holiday-3), min(n_days, holiday+4)):
                holiday_effects[i] += 30 * np.exp(-0.5 * ((i - holiday) ** 2))
    
    # Weekend effects (higher demand on weekends)
    weekend_effects = np.array([20 if date.weekday() >= 5 else 0 for date in dates])
    
    # Random noise with varying volatility
    volatility = 8 + 4 * np.sin(2 * np.pi * np.arange(n_days) / 365.25)  # Seasonal volatility
    noise = np.random.normal(0, volatility)
    
    # Combine all components
    demand = base_demand + yearly_cycle + weekly_cycle + monthly_cycle + holiday_effects + weekend_effects + noise
    demand = np.maximum(demand, 10)  # Ensure minimum demand
    
    # Create DataFrame
    demand_df = pd.DataFrame({
        'ds': dates,
        'y': demand,
        'base_demand': base_demand,
        'yearly_cycle': yearly_cycle,
        'weekly_cycle': weekly_cycle,
        'holiday_effects': holiday_effects,
        'weekend_effects': weekend_effects
    })
    
    return demand_df

def implement_arima_forecasting(ts_data, forecast_periods=30):
    """
    Implement ARIMA forecasting with automatic parameter selection.
    """
    print("🔮 ARIMA Forecasting Implementation...")
    
    # Prepare data
    y = ts_data['y'].values
    
    # Check stationarity
    adf_result = adfuller(y)
    print(f"📊 Stationarity Test (ADF):")
    print(f"  ADF Statistic: {adf_result[0]:.6f}")
    print(f"  p-value: {adf_result[1]:.6f}")
    print(f"  Result: {'Stationary' if adf_result[1] < 0.05 else 'Non-stationary'}")
    
    # Auto ARIMA model selection
    print("🤖 Auto ARIMA model selection...")
    auto_model = pm.auto_arima(y, 
                              start_p=0, start_q=0, 
                              max_p=5, max_q=5, 
                              seasonal=True, 
                              stepwise=True,
                              suppress_warnings=True,
                              error_action='ignore')
    
    print(f"📈 Best ARIMA model: {auto_model.order} x {auto_model.seasonal_order}")
    
    # Fit the model
    model_fit = auto_model
    
    # Make predictions
    forecast, conf_int = model_fit.predict(n_periods=forecast_periods, return_conf_int=True)
    
    # Calculate residuals and diagnostics
    residuals = model_fit.resid()
    
    # Ljung-Box test for residual autocorrelation
    lb_test = acorr_ljungbox(residuals, lags=10, return_df=True)
    
    print(f"🧪 Model Diagnostics:")
    print(f"  AIC: {model_fit.aic():.2f}")
    print(f"  BIC: {model_fit.bic():.2f}")
    print(f"  Ljung-Box p-value: {lb_test['lb_pvalue'].iloc[-1]:.6f}")
    
    return {
        'model': model_fit,
        'forecast': forecast,
        'conf_int': conf_int,
        'residuals': residuals,
        'aic': model_fit.aic(),
        'bic': model_fit.bic()
    }

def implement_prophet_forecasting(ts_data, forecast_periods=30):
    """
    Implement Facebook Prophet forecasting.
    """
    print("🔮 Prophet Forecasting Implementation...")
    
    try:
        from prophet import Prophet
        
        # Prepare data for Prophet
        prophet_data = ts_data[['ds', 'y']].copy()
        
        # Initialize Prophet model with custom parameters
        model = Prophet(
            yearly_seasonality=True,
            weekly_seasonality=True,
            daily_seasonality=False,
            seasonality_mode='additive',
            changepoint_prior_scale=0.05,
            holidays_prior_scale=10.0
        )
        
        # Add custom seasonalities
        model.add_seasonality(name='monthly', period=30.5, fourier_order=5)
        
        # Fit the model
        model.fit(prophet_data)
        
        # Create future dataframe
        future = model.make_future_dataframe(periods=forecast_periods)
        
        # Make predictions
        forecast = model.predict(future)
        
        # Extract metrics
        y_true = prophet_data['y'].values
        y_pred = forecast['yhat'][:len(y_true)].values
        
        mae = mean_absolute_error(y_true, y_pred)
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
        
        print(f"📊 Prophet Model Performance:")
        print(f"  MAE: {mae:.3f}")
        print(f"  RMSE: {rmse:.3f}")
        print(f"  MAPE: {mape:.2f}%")
        
        return {
            'model': model,
            'forecast': forecast,
            'metrics': {'mae': mae, 'rmse': rmse, 'mape': mape}
        }
        
    except ImportError:
        print("⚠️ Prophet not available, using alternative approach...")
        return None

def implement_lstm_forecasting(ts_data, forecast_periods=30, lookback_window=30):
    """
    Implement LSTM neural network for time series forecasting.
    """
    print("🧠 LSTM Neural Network Forecasting...")
    
    try:
        from tensorflow.keras.models import Sequential
        from tensorflow.keras.layers import LSTM, Dense, Dropout
        from tensorflow.keras.optimizers import Adam
        from sklearn.preprocessing import MinMaxScaler
        
        # Prepare data
        data = ts_data['y'].values.reshape(-1, 1)
        
        # Scale data
        scaler = MinMaxScaler()
        scaled_data = scaler.fit_transform(data)
        
        # Create sequences
        def create_sequences(data, lookback):
            X, y = [], []
            for i in range(lookback, len(data)):
                X.append(data[i-lookback:i, 0])
                y.append(data[i, 0])
            return np.array(X), np.array(y)
        
        X, y = create_sequences(scaled_data, lookback_window)
        
        # Split data
        train_size = int(0.8 * len(X))
        X_train, X_test = X[:train_size], X[train_size:]
        y_train, y_test = y[:train_size], y[train_size:]
        
        # Reshape for LSTM
        X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
        X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))
        
        # Build LSTM model
        model = Sequential([
            LSTM(50, return_sequences=True, input_shape=(lookback_window, 1)),
            Dropout(0.2),
            LSTM(50, return_sequences=False),
            Dropout(0.2),
            Dense(25),
            Dense(1)
        ])
        
        model.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
        
        # Train model
        history = model.fit(X_train, y_train, 
                           batch_size=32, epochs=50, 
                           validation_data=(X_test, y_test),
                           verbose=0)
        
        # Make predictions
        predictions = model.predict(X_test)
        predictions = scaler.inverse_transform(predictions)
        y_test_actual = scaler.inverse_transform(y_test.reshape(-1, 1))
        
        # Calculate metrics
        mae = mean_absolute_error(y_test_actual, predictions)
        rmse = np.sqrt(mean_squared_error(y_test_actual, predictions))
        
        print(f"📊 LSTM Model Performance:")
        print(f"  MAE: {mae:.3f}")
        print(f"  RMSE: {rmse:.3f}")
        
        return {
            'model': model,
            'scaler': scaler,
            'history': history,
            'predictions': predictions,
            'metrics': {'mae': mae, 'rmse': rmse}
        }
        
    except ImportError:
        print("⚠️ TensorFlow not available for LSTM implementation")
        return None

# Use the existing demand series data instead of generating new
dates = pd.date_range(start='2022-01-01', periods=len(demand_series), freq='D')
demand_ts = pd.DataFrame({
    'ds': dates,
    'y': demand_series,
    'base_demand': 80 + np.linspace(0, 40, len(demand_series)),
    'yearly_cycle': 25 * np.sin(2 * np.pi * np.arange(len(demand_series)) / 365.25),
    'weekly_cycle': 15 * np.sin(2 * np.pi * np.arange(len(demand_series)) / 7),
    'holiday_effects': np.zeros(len(demand_series)),
    'weekend_effects': np.array([20 if i % 7 >= 5 else 0 for i in range(len(demand_series))])
})

print(f"✅ Using existing demand data: {len(demand_ts)} days")

# Implement ARIMA forecasting with error handling
try:
    arima_results = implement_arima_forecasting(demand_ts, forecast_periods=30)
    print("✅ ARIMA forecasting completed successfully")
except Exception as e:
    print(f"⚠️ ARIMA forecasting failed: {e}")
    arima_results = None

# Implement Prophet forecasting with error handling
try:
    prophet_results = implement_prophet_forecasting(demand_ts, forecast_periods=30)
    print("✅ Prophet forecasting completed successfully")
except Exception as e:
    print(f"⚠️ Prophet forecasting failed (library may not be installed): {e}")
    prophet_results = None

# Implement LSTM forecasting with error handling
try:
    lstm_results = implement_lstm_forecasting(demand_ts, forecast_periods=30)
    print("✅ LSTM forecasting completed successfully")
except Exception as e:
    print(f"⚠️ LSTM forecasting failed (TensorFlow may not be installed): {e}")
    lstm_results = None

# Compare model performances
print("\n🏆 MODEL COMPARISON:")
print("-" * 40)
print(f"ARIMA - AIC: {arima_results['aic']:.2f}, BIC: {arima_results['bic']:.2f}")
if prophet_results:
    print(f"Prophet - MAE: {prophet_results['metrics']['mae']:.3f}, MAPE: {prophet_results['metrics']['mape']:.2f}%")
if lstm_results:
    print(f"LSTM - MAE: {lstm_results['metrics']['mae']:.3f}, RMSE: {lstm_results['metrics']['rmse']:.3f}")

print("\n✅ Time Series Forecasting Complete!")
```

```python
# ============================================================================
# COMPREHENSIVE MACHINE LEARNING PIPELINE
# ============================================================================

print("\n🤖 COMPREHENSIVE MACHINE LEARNING PIPELINE")
print("=" * 60)

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.feature_selection import SelectKBest, f_regression, RFE
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import make_scorer

def create_ml_pipeline(enhanced_data):
    """
    Create comprehensive ML pipeline for delivery time prediction.
    """
    print("🔧 Creating ML Pipeline for Delivery Time Prediction...")
    
    # Define target variable (predict service_time)
    target = 'service_time'
    
    # Select features for modeling
    feature_columns = [
        'package_weight', 'package_volume', 'distance_to_center', 
        'package_density', 'package_complexity', 'time_window_duration',
        'time_flexibility', 'priority_score', 'delivery_difficulty',
        'avg_distance_to_others', 'isolation_score', 'nearest_neighbor_distance',
        'service_efficiency', 'distance_priority_interaction', 
        'weight_volume_interaction', 'time_complexity_interaction'
    ]
    
    # Prepare data
    X = enhanced_data[feature_columns].copy()
    y = enhanced_data[target].copy()
    
    # Handle any missing values
    X = X.fillna(X.mean())
    
    print(f"📊 Dataset shape: {X.shape}")
    print(f"🎯 Target variable: {target}")
    print(f"📋 Features: {len(feature_columns)}")
    
    return X, y, feature_columns

def implement_multiple_models(X, y, test_size=0.2, random_state=42):
    """
    Implement and compare multiple ML models.
    """
    print("🚀 Training Multiple ML Models...")
    
    # Split data
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)
    
    # Define models
    models = {
        'Random Forest': RandomForestRegressor(n_estimators=100, random_state=random_state),
        'Gradient Boosting': GradientBoostingRegressor(n_estimators=100, random_state=random_state),
        'Extra Trees': ExtraTreesRegressor(n_estimators=100, random_state=random_state),
        'Ridge Regression': Ridge(alpha=1.0),
        'Lasso Regression': Lasso(alpha=1.0),
        'Elastic Net': ElasticNet(alpha=1.0, l1_ratio=0.5),
        'SVR': SVR(kernel='rbf', C=1.0),
        'K-Nearest Neighbors': KNeighborsRegressor(n_neighbors=5)
    }
    
    # Train and evaluate models
    model_results = {}
    
    for name, model in models.items():
        print(f"🔄 Training {name}...")
        
        # Create pipeline with scaling
        pipeline = Pipeline([
            ('scaler', StandardScaler()),
            ('model', model)
        ])
        
        # Fit model
        pipeline.fit(X_train, y_train)
        
        # Predictions
        y_pred_train = pipeline.predict(X_train)
        y_pred_test = pipeline.predict(X_test)
        
        # Calculate metrics
        train_mae = mean_absolute_error(y_train, y_pred_train)
        test_mae = mean_absolute_error(y_test, y_pred_test)
        train_rmse = np.sqrt(mean_squared_error(y_train, y_pred_train))
        test_rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
        train_r2 = r2_score(y_train, y_pred_train)
        test_r2 = r2_score(y_test, y_pred_test)
        
        # Cross-validation
        cv_scores = cross_val_score(pipeline, X_train, y_train, cv=5, scoring='neg_mean_absolute_error')
        cv_mae = -cv_scores.mean()
        cv_std = cv_scores.std()
        
        model_results[name] = {
            'pipeline': pipeline,
            'train_mae': train_mae,
            'test_mae': test_mae,
            'train_rmse': train_rmse,
            'test_rmse': test_rmse,
            'train_r2': train_r2,
            'test_r2': test_r2,
            'cv_mae': cv_mae,
            'cv_std': cv_std,
            'predictions_train': y_pred_train,
            'predictions_test': y_pred_test
        }
        
        print(f"  ✅ {name}: Test MAE={test_mae:.3f}, Test R²={test_r2:.3f}, CV MAE={cv_mae:.3f}±{cv_std:.3f}")
    
    return model_results, X_train, X_test, y_train, y_test

def hyperparameter_optimization(X, y, best_model_name, model_results):
    """
    Perform hyperparameter optimization on the best model.
    """
    print(f"\n🎯 Hyperparameter Optimization for {best_model_name}...")
    
    # Define parameter grids
    param_grids = {
        'Random Forest': {
            'model__n_estimators': [50, 100, 200],
            'model__max_depth': [None, 10, 20],
            'model__min_samples_split': [2, 5, 10],
            'model__min_samples_leaf': [1, 2, 4]
        },
        'Gradient Boosting': {
            'model__n_estimators': [50, 100, 200],
            'model__learning_rate': [0.01, 0.1, 0.2],
            'model__max_depth': [3, 5, 7],
            'model__subsample': [0.8, 0.9, 1.0]
        },
        'Ridge Regression': {
            'model__alpha': [0.1, 1.0, 10.0, 100.0]
        },
        'SVR': {
            'model__C': [0.1, 1, 10],
            'model__gamma': ['scale', 'auto', 0.001, 0.01],
            'model__epsilon': [0.01, 0.1, 0.2]
        }
    }
    
    if best_model_name in param_grids:
        # Get the best model pipeline
        best_pipeline = model_results[best_model_name]['pipeline']
        
        # Perform grid search
        grid_search = GridSearchCV(
            best_pipeline,
            param_grids[best_model_name],
            cv=5,
            scoring='neg_mean_absolute_error',
            n_jobs=-1,
            verbose=0
        )
        
        grid_search.fit(X, y)
        
        print(f"🏆 Best parameters: {grid_search.best_params_}")
        print(f"🎯 Best CV score: {-grid_search.best_score_:.3f}")
        
        return grid_search.best_estimator_, grid_search.best_params_
    
    return None, None

def feature_importance_analysis(model_results, feature_columns):
    """
    Analyze feature importance across models.
    """
    print("\n📊 FEATURE IMPORTANCE ANALYSIS")
    print("-" * 40)
    
    # Models that provide feature importance
    importance_models = ['Random Forest', 'Gradient Boosting', 'Extra Trees']
    
    feature_importance_df = pd.DataFrame(index=feature_columns)
    
    for model_name in importance_models:
        if model_name in model_results:
            model = model_results[model_name]['pipeline'].named_steps['model']
            if hasattr(model, 'feature_importances_'):
                feature_importance_df[model_name] = model.feature_importances_
    
    # Calculate average importance
    if len(feature_importance_df.columns) > 0:
        feature_importance_df['Average'] = feature_importance_df.mean(axis=1)
        feature_importance_df = feature_importance_df.sort_values('Average', ascending=False)
        
        print("🔝 Top 10 Most Important Features:")
        for i, (feature, importance) in enumerate(feature_importance_df['Average'].head(10).items()):
            print(f"  {i+1:2d}. {feature}: {importance:.4f}")
        
        return feature_importance_df
    
    return None

def model_interpretation_analysis(best_model, X_test, y_test, feature_columns):
    """
    Perform model interpretation analysis.
    """
    print("\n🔍 MODEL INTERPRETATION ANALYSIS")
    print("-" * 40)
    
    try:
        import shap
        
        # Get the actual model from pipeline
        model = best_model.named_steps['model']
        scaler = best_model.named_steps['scaler']
        
        # Scale the test data
        X_test_scaled = scaler.transform(X_test)
        
        # Create SHAP explainer
        if hasattr(model, 'predict'):
            explainer = shap.Explainer(model, X_test_scaled[:100])  # Use subset for speed
            shap_values = explainer(X_test_scaled[:10])  # Explain first 10 predictions
            
            print("✅ SHAP analysis completed")
            return shap_values
        
    except ImportError:
        print("⚠️ SHAP not available for model interpretation")
    
    return None

# Create ML pipeline
X, y, feature_columns = create_ml_pipeline(enhanced_delivery_data)

# Train multiple models
model_results, X_train, X_test, y_train, y_test = implement_multiple_models(X, y)

# Find best model
best_model_name = min(model_results.keys(), key=lambda k: model_results[k]['test_mae'])
print(f"\n🏆 BEST MODEL: {best_model_name}")
print(f"📊 Test MAE: {model_results[best_model_name]['test_mae']:.3f}")
print(f"📊 Test R²: {model_results[best_model_name]['test_r2']:.3f}")

# Hyperparameter optimization
optimized_model, best_params = hyperparameter_optimization(X, y, best_model_name, model_results)

# Feature importance analysis
feature_importance = feature_importance_analysis(model_results, feature_columns)

# Model interpretation
if optimized_model:
    shap_values = model_interpretation_analysis(optimized_model, X_test, y_test, feature_columns)

# Model comparison summary
print("\n📋 MODEL COMPARISON SUMMARY:")
print("-" * 50)
comparison_df = pd.DataFrame({
    'Model': list(model_results.keys()),
    'Test MAE': [results['test_mae'] for results in model_results.values()],
    'Test R²': [results['test_r2'] for results in model_results.values()],
    'CV MAE': [results['cv_mae'] for results in model_results.values()],
    'CV Std': [results['cv_std'] for results in model_results.values()]
}).round(4)

comparison_df = comparison_df.sort_values('Test MAE')
display(comparison_df)

print("\n✅ Machine Learning Pipeline Complete!")
```

## 3. 📊 Advanced Exploratory Data Analysis & Statistical Testing

### **Multi-Dimensional Data Analysis**
We'll perform comprehensive statistical analysis including:
- **Hypothesis testing** for customer segment differences
- **Correlation analysis** with multiple methods
- **Distribution analysis** and normality testing
- **Time series decomposition** of demand patterns
- **Geospatial clustering** and hotspot analysis


```python
# ============================================================================
# COMPREHENSIVE STATISTICAL ANALYSIS & HYPOTHESIS TESTING
# ============================================================================

def perform_comprehensive_eda(delivery_orders, customers_df, demand_series):
    """
    Perform advanced exploratory data analysis with statistical testing.
    """
    
    print("🔍 COMPREHENSIVE EXPLORATORY DATA ANALYSIS")
    print("=" * 60)
    
    # 1. BASIC DESCRIPTIVE STATISTICS
    print("\n📊 1. DESCRIPTIVE STATISTICS BY CUSTOMER SEGMENT")
    print("-" * 50)
    
    segment_stats = delivery_orders.groupby('customer_segment').agg({
        'package_weight': ['mean', 'std', 'median', 'count'],
        'package_volume': ['mean', 'std', 'median'],
        'order_value': ['mean', 'std', 'median'],
        'service_time': ['mean', 'std', 'median'],
        'customer_loyalty': ['mean', 'std'],
        'time_sensitivity': ['mean', 'std']
    }).round(3)
    
    display(segment_stats)
    
    # 2. STATISTICAL HYPOTHESIS TESTING
    print("\\n🧪 2. HYPOTHESIS TESTING - CUSTOMER SEGMENT DIFFERENCES")
    print("-" * 55)
    
    # Test if order values differ significantly between segments
    high_value_orders = delivery_orders[delivery_orders['customer_segment'] == 'high_value']['order_value']
    medium_value_orders = delivery_orders[delivery_orders['customer_segment'] == 'medium_value']['order_value']
    low_value_orders = delivery_orders[delivery_orders['customer_segment'] == 'low_value']['order_value']
    
    # Shapiro-Wilk test for normality
    shapiro_high = shapiro(high_value_orders.sample(min(50, len(high_value_orders))))
    shapiro_medium = shapiro(medium_value_orders.sample(min(50, len(medium_value_orders))))
    shapiro_low = shapiro(low_value_orders.sample(min(50, len(low_value_orders))))
    
    print(f"Normality Tests (Shapiro-Wilk):")
    print(f"  High Value: p-value = {shapiro_high.pvalue:.6f}")
    print(f"  Medium Value: p-value = {shapiro_medium.pvalue:.6f}")
    print(f"  Low Value: p-value = {shapiro_low.pvalue:.6f}")
    
    # ANOVA test for order value differences
    f_stat, p_value_anova = stats.f_oneway(high_value_orders, medium_value_orders, low_value_orders)
    print(f"\\nANOVA Test (Order Values):")
    print(f"  F-statistic: {f_stat:.4f}")
    print(f"  p-value: {p_value_anova:.6f}")
    print(f"  Result: {'Significant difference' if p_value_anova < 0.05 else 'No significant difference'}")
    
    # Mann-Whitney U test (non-parametric alternative)
    u_stat_hm, p_val_hm = mannwhitneyu(high_value_orders, medium_value_orders)
    u_stat_hl, p_val_hl = mannwhitneyu(high_value_orders, low_value_orders)
    u_stat_ml, p_val_ml = mannwhitneyu(medium_value_orders, low_value_orders)
    
    print(f"\\nMann-Whitney U Tests (pairwise):")
    print(f"  High vs Medium: p-value = {p_val_hm:.6f}")
    print(f"  High vs Low: p-value = {p_val_hl:.6f}")
    print(f"  Medium vs Low: p-value = {p_val_ml:.6f}")
    
    # 3. CORRELATION ANALYSIS
    print("\\n🔗 3. ADVANCED CORRELATION ANALYSIS")
    print("-" * 40)
    
    # Prepare numerical columns for correlation
    numeric_cols = ['package_weight', 'package_volume', 'order_value', 'service_time', 
                   'customer_loyalty', 'time_sensitivity']
    
    corr_data = delivery_orders[numeric_cols]
    
    # Multiple correlation methods
    pearson_corr = corr_data.corr(method='pearson')
    spearman_corr = corr_data.corr(method='spearman')
    kendall_corr = corr_data.corr(method='kendall')
    
    print("Correlation Methods Comparison (Order Value vs Package Weight):")
    print(f"  Pearson: {pearson_corr.loc['order_value', 'package_weight']:.4f}")
    print(f"  Spearman: {spearman_corr.loc['order_value', 'package_weight']:.4f}")
    print(f"  Kendall: {kendall_corr.loc['order_value', 'package_weight']:.4f}")
    
    # 4. TIME SERIES ANALYSIS
    print("\\n📈 4. TIME SERIES DECOMPOSITION")
    print("-" * 35)
    
    # Create time series index
    dates = pd.date_range(start='2024-01-01', periods=len(demand_series), freq='D')
    ts_data = pd.Series(demand_series, index=dates)
    
    # Perform seasonal decomposition
    decomposition = seasonal_decompose(ts_data, model='additive', period=7)  # Weekly seasonality
    
    # Augmented Dickey-Fuller test for stationarity
    adf_result = adfuller(demand_series)
    print(f"Stationarity Test (ADF):")
    print(f"  ADF Statistic: {adf_result[0]:.6f}")
    print(f"  p-value: {adf_result[1]:.6f}")
    print(f"  Critical Values: {adf_result[4]}")
    print(f"  Result: {'Stationary' if adf_result[1] < 0.05 else 'Non-stationary'}")
    
    # 5. GEOGRAPHIC CLUSTERING ANALYSIS
    print("\\n🗺️  5. GEOSPATIAL CLUSTERING ANALYSIS")
    print("-" * 40)
    
    # Prepare coordinates for clustering
    coords = delivery_orders[['latitude', 'longitude']].values
    
    # K-means clustering
    kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(coords)
    
    # Calculate silhouette score
    silhouette_avg = silhouette_score(coords, cluster_labels)
    calinski_score = calinski_harabasz_score(coords, cluster_labels)
    
    print(f"K-means Clustering (k=4):")
    print(f"  Silhouette Score: {silhouette_avg:.4f}")
    print(f"  Calinski-Harabasz Score: {calinski_score:.4f}")
    
    # Add cluster labels to dataframe
    delivery_orders['cluster'] = cluster_labels
    
    # Cluster statistics
    cluster_stats = delivery_orders.groupby('cluster').agg({
        'order_value': ['mean', 'count'],
        'package_weight': 'mean',
        'customer_segment': lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else 'unknown'
    })
    
    print("\\nCluster Characteristics:")
    display(cluster_stats)
    
    return {
        'segment_stats': segment_stats,
        'hypothesis_tests': {
            'anova': {'f_stat': f_stat, 'p_value': p_value_anova},
            'normality': {'high': shapiro_high, 'medium': shapiro_medium, 'low': shapiro_low},
            'mannwhitney': {'hm': p_val_hm, 'hl': p_val_hl, 'ml': p_val_ml}
        },
        'correlations': {'pearson': pearson_corr, 'spearman': spearman_corr, 'kendall': kendall_corr},
        'time_series': {'decomposition': decomposition, 'adf_test': adf_result},
        'clustering': {'silhouette': silhouette_avg, 'calinski': calinski_score, 'labels': cluster_labels}
    }

# Execute comprehensive EDA with proper data
print("🔍 Executing comprehensive statistical analysis...")

# First, let's create some sample data for demonstration
print("📊 Preparing data for analysis...")

# Create sample data that matches the expected format
sample_delivery_orders = delivery_orders.copy()

# Perform comprehensive analysis
analysis_results = perform_comprehensive_eda(sample_delivery_orders, customers_df, demand_series)

print("✅ Statistical Analysis Complete!")
```

```python
# ============================================================================
# COMPREHENSIVE BUSINESS IMPACT ANALYSIS
# ============================================================================

print("💰 BUSINESS IMPACT & ROI ANALYSIS")
print("=" * 45)

# Calculate naive baseline (simple sequential routing)
def calculate_naive_baseline():
    """Calculate baseline performance using simple sequential routing."""
    naive_distance = 0
    naive_time = 0
    naive_cost = 0
    
    points_per_vehicle = len(delivery_points) // len(vehicles)
    
    for i, vehicle in vehicles.iterrows():
        start_idx = i * points_per_vehicle
        end_idx = min((i + 1) * points_per_vehicle, len(delivery_points))
        
        if start_idx < len(delivery_points):
            route_distance = 0
            for j in range(start_idx, min(end_idx - 1, len(delivery_points) - 1)):
                route_distance += distance_matrix[j][j + 1]
            
            # Add return to depot
            if end_idx - 1 < len(delivery_points):
                route_distance += distance_matrix[end_idx - 1][0]
            
            naive_distance += route_distance
            naive_time += (route_distance / vehicle['avg_speed']) * 60
            naive_cost += route_distance * vehicle['cost_per_km']
    
    return naive_distance, naive_time, naive_cost

print("📊 Calculating baseline performance (naive routing)...")
naive_distance, naive_time, naive_cost = calculate_naive_baseline()

# Calculate savings from optimization
distance_savings = naive_distance - metrics_ortools['total_distance']
time_savings = naive_time - metrics_ortools['total_time']
cost_savings = naive_cost - metrics_ortools['total_cost']

distance_savings_pct = (distance_savings / naive_distance) * 100
time_savings_pct = (time_savings / naive_time) * 100
cost_savings_pct = (cost_savings / naive_cost) * 100

print("✅ Baseline calculation complete!")

# Create comparison DataFrame
comparison_results = pd.DataFrame({
    'Metric': ['Total Distance (km)', 'Total Time (min)', 'Total Cost ($)', 'Vehicles Used'],
    'Naive Routing': [naive_distance, naive_time, naive_cost, len(vehicles)],
    'OR-Tools Optimized': [metrics_ortools['total_distance'], metrics_ortools['total_time'], 
                          metrics_ortools['total_cost'], len(routes_ortools)],
    'Absolute Savings': [distance_savings, time_savings, cost_savings, 0],
    'Percentage Improvement': [distance_savings_pct, time_savings_pct, cost_savings_pct, 0]
})

print("\\n📈 OPTIMIZATION PERFORMANCE COMPARISON")
print("-" * 50)
display(comparison_results.round(2))

# Annual projections
working_days_per_year = 250
annual_distance_savings = distance_savings * working_days_per_year
annual_time_savings_hours = (time_savings * working_days_per_year) / 60
annual_cost_savings = cost_savings * working_days_per_year

# Environmental impact
co2_per_km = 0.2  # kg CO2 per km (average delivery vehicle)
annual_co2_reduction = annual_distance_savings * co2_per_km

print(f"\\n🎯 KEY PERFORMANCE IMPROVEMENTS")
print("-" * 40)
print(f"Distance Reduction: {distance_savings:.1f} km ({distance_savings_pct:.1f}%)")
print(f"Time Savings: {time_savings:.0f} minutes ({time_savings_pct:.1f}%)")
print(f"Cost Reduction: ${cost_savings:.2f} ({cost_savings_pct:.1f}%)")

print(f"\\n📅 ANNUAL IMPACT PROJECTIONS (250 working days)")
print("-" * 50)
print(f"Distance Savings: {annual_distance_savings:,.0f} km")
print(f"Time Savings: {annual_time_savings_hours:,.0f} hours")
print(f"Cost Savings: ${annual_cost_savings:,.2f}")
print(f"CO2 Reduction: {annual_co2_reduction:,.0f} kg")

# ROI Analysis
implementation_cost = 50000  # Estimated cost to implement optimization system
monthly_savings = annual_cost_savings / 12
payback_period_months = implementation_cost / monthly_savings
annual_roi = (annual_cost_savings / implementation_cost) * 100

print(f"\\n💵 RETURN ON INVESTMENT ANALYSIS")
print("-" * 40)
print(f"Estimated Implementation Cost: ${implementation_cost:,.2f}")
print(f"Monthly Cost Savings: ${monthly_savings:,.2f}")
print(f"Payback Period: {payback_period_months:.1f} months")
print(f"Annual ROI: {annual_roi:.1f}%")

# Create business impact visualization
fig_impact = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Cost Comparison', 'Time Savings', 'Distance Reduction', 'Annual Projections'),
    specs=[[{"type": "bar"}, {"type": "bar"}],
           [{"type": "bar"}, {"type": "bar"}]]
)

# Cost comparison
methods = ['Naive Routing', 'OR-Tools Optimized']
costs = [naive_cost, metrics_ortools['total_cost']]
fig_impact.add_trace(go.Bar(x=methods, y=costs, name='Daily Cost', marker_color=['red', 'green']), row=1, col=1)

# Time comparison  
times = [naive_time, metrics_ortools['total_time']]
fig_impact.add_trace(go.Bar(x=methods, y=times, name='Daily Time', marker_color=['red', 'green']), row=1, col=2)

# Distance comparison
distances = [naive_distance, metrics_ortools['total_distance']]
fig_impact.add_trace(go.Bar(x=methods, y=distances, name='Daily Distance', marker_color=['red', 'green']), row=2, col=1)

# Annual savings
savings_categories = ['Cost Savings', 'CO2 Reduction', 'Time Savings (hrs)']
savings_values = [annual_cost_savings, annual_co2_reduction, annual_time_savings_hours]
fig_impact.add_trace(go.Bar(x=savings_categories, y=savings_values, name='Annual Savings', 
                           marker_color=['blue', 'green', 'orange']), row=2, col=2)

fig_impact.update_layout(
    title='Business Impact Analysis - Route Optimization ROI',
    height=700,
    showlegend=False
)

print("\\n📊 Displaying Business Impact Dashboard:")
fig_impact.show()

print("\\n✅ Business Impact Analysis Complete!")
print("🎉 Route optimization demonstrates significant value creation!")
```
